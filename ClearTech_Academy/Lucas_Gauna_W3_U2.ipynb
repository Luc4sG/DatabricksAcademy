**Se crea un data frame a partir de una lista de tuplas de python.**


```sql
%python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType

# Define schema for livestock registry
schema = StructType([
    StructField("animal_id", StringType(), True),
    StructField("species", StringType(), True),
    StructField("breed", StringType(), True),
    StructField("age_years", IntegerType(), True),
    StructField("weight_kg", DoubleType(), True),
    StructField("owner_name", StringType(), True),
    StructField("farm_id", StringType(), True)
])

# Example data: list of tuples (each representing an animal)
data = [
    ("A001", "Cattle", "Angus", 4, 620.5, "Juan Pérez", "F001"),
    ("A002", "Cattle", "Hereford", 2, 540.2, "Juan Pérez", "F001"),
    ("A003", "Sheep", "Merino", 3, 72.8, "María López", "F002"),
    ("A004", "Pig", "Duroc", 1, 110.0, "Carlos Gómez", "F003"),
    ("A005", "Cattle", "Holstein", 5, 655.4, "Ana Torres", "F004")
]

df_livestock = spark.createDataFrame(data, schema)
df_livestock.display()
```
