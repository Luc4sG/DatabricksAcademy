```python
%sql
-- CATÁLOGOS
CREATE CATALOG IF NOT EXISTS dev;
CREATE CATALOG IF NOT EXISTS prod;

-- ESQUEMAS - DEV
CREATE SCHEMA IF NOT EXISTS dev.sales;
CREATE SCHEMA IF NOT EXISTS dev.marketing;
CREATE SCHEMA IF NOT EXISTS dev.hr;

-- ESQUEMAS - PROD
CREATE SCHEMA IF NOT EXISTS prod.sales;
CREATE SCHEMA IF NOT EXISTS prod.marketing;
CREATE SCHEMA IF NOT EXISTS prod.hr;
```

```python
%sql
-- TABLAS - DEV.SALES
CREATE TABLE IF NOT EXISTS dev.sales.website_sales (
    sale_id BIGINT,
    customer_id BIGINT,
    product_id BIGINT,
    sale_date DATE,
    amount DECIMAL(10,2),
    created_at TIMESTAMP
);

CREATE TABLE IF NOT EXISTS dev.sales.retail_sales (
    sale_id BIGINT,
    store_id BIGINT,
    product_id BIGINT,
    sale_date DATE,
    amount DECIMAL(10,2)
);

-- TABLAS - PROD.SALES
CREATE TABLE IF NOT EXISTS prod.sales.website_sales (
    sale_id BIGINT,
    customer_id BIGINT,
    product_id BIGINT,
    sale_date DATE,
    amount DECIMAL(10,2),
    created_at TIMESTAMP
);

CREATE TABLE IF NOT EXISTS prod.sales.retail_sales (
    sale_id BIGINT,
    store_id BIGINT,
    product_id BIGINT,
    sale_date DATE,
    amount DECIMAL(10,2)
);

-- TABLAS - DEV.MARKETING
CREATE TABLE IF NOT EXISTS dev.marketing.campaigns (
    campaign_id BIGINT,
    campaign_name STRING,
    start_date DATE,
    budget DECIMAL(10,2)
);

CREATE TABLE IF NOT EXISTS dev.marketing.leads (
    lead_id BIGINT,
    source STRING,
    created_date DATE,
    status STRING
);

-- TABLAS - PROD.MARKETING
CREATE TABLE IF NOT EXISTS prod.marketing.campaigns (
    campaign_id BIGINT,
    campaign_name STRING,
    start_date DATE,
    budget DECIMAL(10,2)
);

CREATE TABLE IF NOT EXISTS prod.marketing.leads (
    lead_id BIGINT,
    source STRING,
    created_date DATE,
    status STRING
);

-- TABLAS - DEV.HR
CREATE TABLE IF NOT EXISTS dev.hr.employees (
    employee_id BIGINT,
    name STRING,
    department STRING,
    hire_date DATE
);

CREATE TABLE IF NOT EXISTS dev.hr.payroll (
    employee_id BIGINT,
    salary DECIMAL(10,2),
    payment_date DATE
);

-- TABLAS - PROD.HR
CREATE TABLE IF NOT EXISTS prod.hr.employees (
    employee_id BIGINT,
    name STRING,
    department STRING,
    hire_date DATE
);

CREATE TABLE IF NOT EXISTS prod.hr.payroll (
    employee_id BIGINT,
    salary DECIMAL(10,2),
    payment_date DATE
);
```

```python
%sql
-- CREACIÓN DE GRUPOS
CREATE GROUP IF NOT EXISTS online_sales;
CREATE GROUP IF NOT EXISTS sales;
CREATE GROUP IF NOT EXISTS marketing;
CREATE GROUP IF NOT EXISTS hr;
CREATE GROUP IF NOT EXISTS ti_admins;
CREATE GROUP IF NOT EXISTS ti_devs;
```

```python
%sql
-- PERMISOS: ONLINE_SALES
-- Solo lectura en website_sales de ambos ambientes
GRANT USAGE ON CATALOG dev TO online_sales;
GRANT USAGE ON CATALOG prod TO online_sales;
GRANT USAGE ON SCHEMA dev.sales TO online_sales;
GRANT USAGE ON SCHEMA prod.sales TO online_sales;
GRANT SELECT ON TABLE dev.sales.website_sales TO online_sales;
GRANT SELECT ON TABLE prod.sales.website_sales TO online_sales;

-- PERMISOS: SALES
-- Control total en esquemas sales de ambos ambientes
GRANT USAGE ON CATALOG dev TO sales;
GRANT USAGE ON CATALOG prod TO sales;
GRANT ALL PRIVILEGES ON SCHEMA dev.sales TO sales;
GRANT ALL PRIVILEGES ON SCHEMA prod.sales TO sales;

-- PERMISOS: MARKETING
-- Control total en esquemas marketing de ambos ambientes
GRANT USAGE ON CATALOG dev TO marketing;
GRANT USAGE ON CATALOG prod TO marketing;
GRANT ALL PRIVILEGES ON SCHEMA dev.marketing TO marketing;
GRANT ALL PRIVILEGES ON SCHEMA prod.marketing TO marketing;

-- PERMISOS: HR
-- Control total en esquemas hr de ambos ambientes
GRANT USAGE ON CATALOG dev TO hr;
GRANT USAGE ON CATALOG prod TO hr;
GRANT ALL PRIVILEGES ON SCHEMA dev.hr TO hr;
GRANT ALL PRIVILEGES ON SCHEMA prod.hr TO hr;

-- PERMISOS: TI_ADMINS
-- Crear/borrar objetos en todos los esquemas SIN acceso a datos
GRANT USAGE, CREATE SCHEMA ON CATALOG dev TO ti_admins;
GRANT USAGE, CREATE SCHEMA ON CATALOG prod TO ti_admins;
GRANT USAGE, CREATE, DROP ON SCHEMA dev.sales TO ti_admins;
GRANT USAGE, CREATE, DROP ON SCHEMA dev.marketing TO ti_admins;
GRANT USAGE, CREATE, DROP ON SCHEMA dev.hr TO ti_admins;
GRANT USAGE, CREATE, DROP ON SCHEMA prod.sales TO ti_admins;
GRANT USAGE, CREATE, DROP ON SCHEMA prod.marketing TO ti_admins;
GRANT USAGE, CREATE, DROP ON SCHEMA prod.hr TO ti_admins;

-- PERMISOS: TI_DEVS
-- Lectura en todos los esquemas de ambos ambientes
GRANT USAGE ON CATALOG dev TO ti_devs;
GRANT USAGE ON CATALOG prod TO ti_devs;
GRANT USAGE, SELECT ON SCHEMA dev.sales TO ti_devs;
GRANT USAGE, SELECT ON SCHEMA dev.marketing TO ti_devs;
GRANT USAGE, SELECT ON SCHEMA dev.hr TO ti_devs;
GRANT USAGE, SELECT ON SCHEMA prod.sales TO ti_devs;
GRANT USAGE, SELECT ON SCHEMA prod.marketing TO ti_devs;
GRANT USAGE, SELECT ON SCHEMA prod.hr TO ti_devs;
```
